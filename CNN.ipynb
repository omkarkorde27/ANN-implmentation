## Q1. What is the use of padding or why padding is used?

### -> Padding is used to get back the original size of the image. So for eg, if we have a 6x6 image and we are using a filter of size 3x3, stride 1 and add 1 layer of padding then the output we get will be of size 6x6. If padding was not used then the output image would have been of size 4x4 resulting in loss of information. 
### Formula to find out the ouput size without padding = n-f+1
### Formula to find out the ouput size with padding = n+2p-f+1/s
